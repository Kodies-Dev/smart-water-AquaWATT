# Porównanie wariantów modeli
Najpierw uruchom `python -m model.eksperymenty`.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import Image, display

W = Path("wyniki")
met = pd.read_csv(W / "metryki.csv")
```

## WAPE dobowe [%] – wariant × obiekt (mniej = lepiej)

```python
tab = met.pivot(index="wariant", columns="obiekt", values="WAPE_doba")
tab = tab[[c for c in tab if c != "RAZEM"] + ["RAZEM"]].sort_values("RAZEM")
tab.round(1).style.background_gradient(cmap="RdYlGn_r", axis=None)
```

## Poprawa względem naiwnego „tydzień temu” [%]

```python
met.pivot(index="wariant", columns="obiekt", values="poprawa_vs_naiwny_proc").round(1).sort_values("RAZEM", ascending=False)
```

## Rybnik – miasto testowe

```python
ry = met[met["obiekt"].str.startswith("RY")].pivot(index="wariant", columns="obiekt", values="WAPE_doba")
ry.sort_values(ry.columns[-1]).plot.barh(figsize=(10, 6), title="WAPE dobowe [%] – Rybnik")
plt.show()
```
